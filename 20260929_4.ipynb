```python
height = float(input("請輸入身高（公分）："))
weight = float(input("請輸入體重（公斤）："))
bmi = weight / ( height/100 )**2
if bmi < 18.5:
  result = "過輕"
elif bmi >= 24:
  result = "過重"
else :
  result = "正常"

print(f"你的 BMI 值為 ：{bmi:.2f}，判定為 ：{result}")
```
