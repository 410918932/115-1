```python
bag = set()
print(f"撿到道具後的背包: {bag}")

bag.add("劍")
bag.add("藥水")
bag.add("盾")
print(f"撿到道具後的背包: {bag}")

bag.add("藥水")
print(f"再次撿到藥水後的背包: {bag}")

bag.discard("盾")
print(f"使用盾後的背包: {bag}")

print(f"掉落的道具: {bag.pop()}")
print(f"遊戲結束，最後背包: {bag}")
```
