```python
import datetime
import traceback
try :
  x = int(input("請輸入被除數: "))
  y = int(input("請輸入除數: "))
  result = x / y
  print(f"x / y = {result}")
except Exception as e:
  print("發生錯誤")
  with open('log_test.txt', 'a', encoding='utf-8') as f:
    f.write(f"{datetime.datetime.now()},發生錯誤: {e}")
    f.write(traceback.format_exc())
finally:
  print('程式結束')
```
