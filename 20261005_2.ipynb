```python
import random
day = 1
win = False
while day<=3 and win!=True:
  lucky_num = random.randint(1,50)
  num = int(input(f"今天第{day}天,請輸入1-50數字"))
  print(f"今天的幸運數字是{lucky_num}")
  if num == lucky_num:
    print("恭喜你中獎")
    win = True
  else:
    print("你沒中獎，明天再來")
  day+=1
if day > 3 and win!=True:
  print("很遺憾3天都沒中獎，下次再試試吧！")
```
