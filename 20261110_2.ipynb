```python
menu = {"蘋果": 50, "香蕉": 30, "牛奶": 60, "麵包": 40}
print("=== 超市商品 ===")
for i,j in menu.items():
  print(i,":",j,"元")
total = 0
while True:
  text = input("請輸入商品名稱（輸入 / 結束）：")
  if text == "/":
    break
  if text in menu:
    total+=menu.get(text)
  else:
    print("查無此商品")
print(f"總金額:{total}元")
```
