```python
box_a=[]
box_b=[]

while True :
  item = input("請輸入A倉文具，輸入'/'結束：")

  if item == "/":
    break
  box_a.append(item)
while True :
  item = input("請輸入B倉文具，輸入'/'結束：")

  if item == "/":
    break
  box_b.append(item)
box = box_a+box_b
box.append("膠水")
box = "/".join(box)
print(f"二倉合併清單：{box}")
```
