```python
weapon =  ["木棒", "手槍", "霰彈槍", "狙擊槍", "雷射劍"]
attack = (10, 30, 50, 80, 100)
avg_attack=0

for i,j in enumerate(weapon,start=1):
  print(i,j)

while True :
  num  = int(input("請輸入武器編號（0 結束）:"))
  if num == 0:
    break
  if 1<=num<=len(weapon):
    print(f"{weapon[num-1]}的攻擊力是{attack[num-1]}")
  else:
    print("找不到武器")
print(f"平均攻擊力：{sum(attack)/len(weapon)}")
```
