```python
def BMI (height,weight):


  value = weight/(height/100)**2

  return value
bmi1 = BMI(170,60)
bmi2 = BMI(165,55)
bmi3 = BMI(180,80)

print(f"第一位{bmi1:.2f}\n第二位{bmi2:.2f}\n第三位{bmi3:.2f}")
```
