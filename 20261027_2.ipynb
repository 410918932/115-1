```python
menu=[]

while True :
  item = input("請輸入您要點的餐點，輸入'/'結束：")




  if item == "/":
    menu.append("冰淇淋")
    break
  menu.append(item)
  if item == "沙拉":
    menu.remove("沙拉")
    print("今日無沙拉，已自動移除。")

order = ",".join(menu)
print(f"完整訂單清單：{order}")
```
