```python
import random

while True:
  ran = random.randint(1,5)
  if ran == 3:
    print("陷阱跳過")
    continue
  elif ran == 5:
    print("出口結束")
    break
  elif ran == 1:
    print("遇到怪物")
  elif ran == 2:
    print("找到寶物")
  elif ran == 4:
    print("補充生命")
```
